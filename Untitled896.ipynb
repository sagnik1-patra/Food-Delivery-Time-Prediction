```python
# ============================================================
# Food Delivery Time Prediction - FINAL ERROR-FREE CODE
# ============================================================

import os
import json
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from math import radians, sin, cos, sqrt, atan2

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression, LogisticRegression
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    mean_squared_error,
    mean_absolute_error,
    r2_score,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix,
    classification_report,
    roc_curve,
    auc
)

# ============================================================
# 1. PATHS
# ============================================================

BASE_DIR = r"C:\Users\NXTWAVE\Downloads\Food Delivery Time Prediction"
DATA_PATH = r"C:\Users\NXTWAVE\Downloads\Food Delivery Time Prediction\Food_Delivery_Time_Prediction.csv"

GRAPH_DIR = os.path.join(BASE_DIR, "graphs")
MODEL_DIR = os.path.join(BASE_DIR, "models")
REPORT_DIR = os.path.join(BASE_DIR, "reports")
RESULT_DIR = os.path.join(BASE_DIR, "results")

for folder in [BASE_DIR, GRAPH_DIR, MODEL_DIR, REPORT_DIR, RESULT_DIR]:
    os.makedirs(folder, exist_ok=True)

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATA_PATH)

print("Dataset Loaded Successfully")
print("Shape:", df.shape)
display(df.head())

df.head().to_csv(os.path.join(RESULT_DIR, "dataset_head.csv"), index=False)

# ============================================================
# 3. BASIC CLEANING
# ============================================================

df.isnull().sum().to_csv(os.path.join(RESULT_DIR, "initial_missing_values.csv"))

numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns
categorical_cols = df.select_dtypes(include=["object"]).columns

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())

for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum().to_csv(os.path.join(RESULT_DIR, "cleaned_missing_values.csv"))

# ============================================================
# 4. LOCATION FEATURE ENGINEERING
# ============================================================

def extract_lat_lon(location):
    try:
        location = str(location).replace("(", "").replace(")", "")
        lat, lon = location.split(",")
        return float(lat.strip()), float(lon.strip())
    except:
        return np.nan, np.nan

df[["Customer_Latitude", "Customer_Longitude"]] = df["Customer_Location"].apply(
    lambda x: pd.Series(extract_lat_lon(x))
)

df[["Restaurant_Latitude", "Restaurant_Longitude"]] = df["Restaurant_Location"].apply(
    lambda x: pd.Series(extract_lat_lon(x))
)

def haversine(lat1, lon1, lat2, lon2):
    try:
        R = 6371
        lat1, lon1, lat2, lon2 = map(radians, [lat1, lon1, lat2, lon2])
        dlat = lat2 - lat1
        dlon = lon2 - lon1

        a = sin(dlat / 2) ** 2 + cos(lat1) * cos(lat2) * sin(dlon / 2) ** 2
        c = 2 * atan2(sqrt(a), sqrt(1 - a))

        return R * c
    except:
        return np.nan

df["Calculated_Distance"] = df.apply(
    lambda row: haversine(
        row["Customer_Latitude"],
        row["Customer_Longitude"],
        row["Restaurant_Latitude"],
        row["Restaurant_Longitude"]
    ),
    axis=1
)

df["Distance"] = df["Distance"].fillna(df["Calculated_Distance"])

# ============================================================
# 5. TIME FEATURE ENGINEERING
# ============================================================

time_map = {
    "Morning": 9,
    "Afternoon": 14,
    "Evening": 19,
    "Night": 22
}

df["Order_Hour"] = df["Order_Time"].map(time_map)

df["Rush_Hour"] = df["Order_Time"].apply(
    lambda x: 1 if x in ["Afternoon", "Evening", "Night"] else 0
)

# ============================================================
# 6. DELIVERY STATUS FOR LOGISTIC REGRESSION
# ============================================================

median_delivery_time = df["Delivery_Time"].median()

df["Delivery_Status"] = df["Delivery_Time"].apply(
    lambda x: "Delayed" if x > median_delivery_time else "Fast"
)

df["Delivery_Status_Binary"] = df["Delivery_Status"].map({
    "Fast": 0,
    "Delayed": 1
})

df.to_csv(os.path.join(RESULT_DIR, "processed_dataset.csv"), index=False)

# ============================================================
# 7. DESCRIPTIVE STATISTICS
# ============================================================

desc_stats = df.describe().T
desc_stats["median"] = df.select_dtypes(include=["int64", "float64"]).median()
desc_stats["variance"] = df.select_dtypes(include=["int64", "float64"]).var()

display(desc_stats)

desc_stats.to_csv(os.path.join(RESULT_DIR, "descriptive_statistics.csv"))

# ============================================================
# 8. EDA VISUALIZATIONS
# ============================================================

plt.figure(figsize=(8, 5))
plt.hist(df["Delivery_Time"].dropna(), bins=30)
plt.title("Distribution of Delivery Time")
plt.xlabel("Delivery Time")
plt.ylabel("Frequency")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "delivery_time_distribution.png"))
plt.show()

plt.figure(figsize=(14, 10))
corr = df.select_dtypes(include=["int64", "float64"]).corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "correlation_heatmap.png"))
plt.show()

corr.to_csv(os.path.join(RESULT_DIR, "correlation_matrix.csv"))

# Safe Boxplots
for col in df.select_dtypes(include=["int64", "float64"]).columns:
    data = pd.to_numeric(df[col], errors="coerce")
    data = data.replace([np.inf, -np.inf], np.nan).dropna()

    if len(data) == 0:
        continue

    plt.figure(figsize=(8, 4))
    plt.boxplot(data, vert=False)
    plt.title(f"Boxplot of {col}")
    plt.xlabel(col)
    plt.tight_layout()
    plt.savefig(os.path.join(GRAPH_DIR, f"boxplot_{col}.png"))
    plt.show()

# ============================================================
# 9. OUTLIER HANDLING
# ============================================================

df_clean = df.copy()

for col in df_clean.select_dtypes(include=["int64", "float64"]).columns:
    if col != "Delivery_Status_Binary":
        data = pd.to_numeric(df_clean[col], errors="coerce")

        Q1 = data.quantile(0.25)
        Q3 = data.quantile(0.75)
        IQR = Q3 - Q1

        lower = Q1 - 1.5 * IQR
        upper = Q3 + 1.5 * IQR

        df_clean[col] = np.where(data < lower, lower, data)
        df_clean[col] = np.where(df_clean[col] > upper, upper, df_clean[col])

df = df_clean.copy()

df.to_csv(os.path.join(RESULT_DIR, "outlier_handled_dataset.csv"), index=False)

# ============================================================
# 10. ENCODING
# ============================================================

drop_cols = [
    "Order_ID",
    "Customer_Location",
    "Restaurant_Location",
    "Delivery_Status"
]

model_df = df.drop(columns=[col for col in drop_cols if col in df.columns])

categorical_cols = model_df.select_dtypes(include=["object"]).columns.tolist()

model_df = pd.get_dummies(model_df, columns=categorical_cols, drop_first=True)

for col in model_df.columns:
    model_df[col] = pd.to_numeric(model_df[col], errors="coerce")

model_df = model_df.replace([np.inf, -np.inf], np.nan)

# ============================================================
# IMPORTANT FIX FOR IMPUTER ERROR
# Remove columns that are completely NaN before imputation
# ============================================================

all_nan_cols = model_df.columns[model_df.isnull().all()].tolist()

if len(all_nan_cols) > 0:
    print("Dropping completely empty columns:", all_nan_cols)
    model_df = model_df.drop(columns=all_nan_cols)

final_columns = model_df.columns.tolist()

final_imputer = SimpleImputer(strategy="median")

model_df = pd.DataFrame(
    final_imputer.fit_transform(model_df),
    columns=final_columns
)

print("Final Dataset Shape:", model_df.shape)
print("Final NaN Count:", model_df.isnull().sum().sum())

model_df.to_csv(os.path.join(RESULT_DIR, "encoded_final_model_dataset.csv"), index=False)

joblib.dump(final_imputer, os.path.join(MODEL_DIR, "final_imputer.pkl"))

# ============================================================
# 11. LINEAR REGRESSION
# ============================================================

X_reg = model_df.drop(columns=["Delivery_Time", "Delivery_Status_Binary"])
y_reg = model_df["Delivery_Time"]

X_train_reg, X_test_reg, y_train_reg, y_test_reg = train_test_split(
    X_reg,
    y_reg,
    test_size=0.2,
    random_state=42
)

scaler_reg = StandardScaler()

X_train_reg_scaled = scaler_reg.fit_transform(X_train_reg)
X_test_reg_scaled = scaler_reg.transform(X_test_reg)

linear_model = LinearRegression()
linear_model.fit(X_train_reg_scaled, y_train_reg)

y_pred_reg = linear_model.predict(X_test_reg_scaled)

mse = mean_squared_error(y_test_reg, y_pred_reg)
rmse = np.sqrt(mse)
mae = mean_absolute_error(y_test_reg, y_pred_reg)
r2 = r2_score(y_test_reg, y_pred_reg)

linear_results = {
    "Model": "Linear Regression",
    "MSE": mse,
    "RMSE": rmse,
    "MAE": mae,
    "R2 Score": r2
}

print("\nLinear Regression Results:")
print(linear_results)

pd.DataFrame([linear_results]).to_csv(
    os.path.join(RESULT_DIR, "linear_regression_results.csv"),
    index=False
)

joblib.dump(linear_model, os.path.join(MODEL_DIR, "linear_regression_model.pkl"))
joblib.dump(scaler_reg, os.path.join(MODEL_DIR, "linear_regression_scaler.pkl"))

plt.figure(figsize=(8, 6))
plt.scatter(y_test_reg, y_pred_reg)
plt.xlabel("Actual Delivery Time")
plt.ylabel("Predicted Delivery Time")
plt.title("Actual vs Predicted Delivery Time")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "linear_actual_vs_predicted.png"))
plt.show()

pd.DataFrame({
    "Actual_Delivery_Time": y_test_reg.values,
    "Predicted_Delivery_Time": y_pred_reg
}).to_csv(
    os.path.join(RESULT_DIR, "linear_regression_predictions.csv"),
    index=False
)

# ============================================================
# 12. LOGISTIC REGRESSION
# ============================================================

X_cls = model_df.drop(columns=["Delivery_Time", "Delivery_Status_Binary"])
y_cls = model_df["Delivery_Status_Binary"]

X_train_cls, X_test_cls, y_train_cls, y_test_cls = train_test_split(
    X_cls,
    y_cls,
    test_size=0.2,
    random_state=42,
    stratify=y_cls
)

scaler_cls = StandardScaler()

X_train_cls_scaled = scaler_cls.fit_transform(X_train_cls)
X_test_cls_scaled = scaler_cls.transform(X_test_cls)

logistic_model = LogisticRegression(max_iter=1000)
logistic_model.fit(X_train_cls_scaled, y_train_cls)

y_pred_cls = logistic_model.predict(X_test_cls_scaled)
y_prob_cls = logistic_model.predict_proba(X_test_cls_scaled)[:, 1]

accuracy = accuracy_score(y_test_cls, y_pred_cls)
precision = precision_score(y_test_cls, y_pred_cls)
recall = recall_score(y_test_cls, y_pred_cls)
f1 = f1_score(y_test_cls, y_pred_cls)

logistic_results = {
    "Model": "Logistic Regression",
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1
}

print("\nLogistic Regression Results:")
print(logistic_results)

print("\nClassification Report:")
print(classification_report(y_test_cls, y_pred_cls))

pd.DataFrame([logistic_results]).to_csv(
    os.path.join(RESULT_DIR, "logistic_regression_results.csv"),
    index=False
)

with open(os.path.join(REPORT_DIR, "classification_report.txt"), "w") as f:
    f.write(classification_report(y_test_cls, y_pred_cls))

joblib.dump(logistic_model, os.path.join(MODEL_DIR, "logistic_regression_model.pkl"))
joblib.dump(scaler_cls, os.path.join(MODEL_DIR, "logistic_regression_scaler.pkl"))

# ============================================================
# 13. CONFUSION MATRIX
# ============================================================

cm = confusion_matrix(y_test_cls, y_pred_cls)

plt.figure(figsize=(6, 5))
sns.heatmap(
    cm,
    annot=True,
    fmt="d",
    cmap="Blues",
    xticklabels=["Fast", "Delayed"],
    yticklabels=["Fast", "Delayed"]
)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Logistic Regression Confusion Matrix")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "logistic_confusion_matrix.png"))
plt.show()

pd.DataFrame(
    cm,
    index=["Actual_Fast", "Actual_Delayed"],
    columns=["Predicted_Fast", "Predicted_Delayed"]
).to_csv(os.path.join(RESULT_DIR, "confusion_matrix.csv"))

# ============================================================
# 14. ROC CURVE
# ============================================================

fpr, tpr, thresholds = roc_curve(y_test_cls, y_prob_cls)
roc_auc = auc(fpr, tpr)

plt.figure(figsize=(8, 6))
plt.plot(fpr, tpr, label=f"ROC AUC = {roc_auc:.2f}")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "logistic_roc_curve.png"))
plt.show()

pd.DataFrame({
    "False_Positive_Rate": fpr,
    "True_Positive_Rate": tpr,
    "Threshold": thresholds
}).to_csv(os.path.join(RESULT_DIR, "roc_curve_data.csv"), index=False)

pd.DataFrame({
    "Actual_Status": y_test_cls.values,
    "Predicted_Status": y_pred_cls,
    "Delayed_Probability": y_prob_cls
}).to_csv(
    os.path.join(RESULT_DIR, "logistic_regression_predictions.csv"),
    index=False
)

# ============================================================
# 15. FEATURE COEFFICIENTS
# ============================================================

linear_features = pd.DataFrame({
    "Feature": X_reg.columns,
    "Coefficient": linear_model.coef_
}).sort_values(by="Coefficient", ascending=False)

linear_features.to_csv(
    os.path.join(RESULT_DIR, "linear_feature_coefficients.csv"),
    index=False
)

plt.figure(figsize=(10, 8))
sns.barplot(data=linear_features.head(15), x="Coefficient", y="Feature")
plt.title("Top Linear Regression Feature Coefficients")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "linear_feature_coefficients.png"))
plt.show()

logistic_features = pd.DataFrame({
    "Feature": X_cls.columns,
    "Coefficient": logistic_model.coef_[0]
}).sort_values(by="Coefficient", ascending=False)

logistic_features.to_csv(
    os.path.join(RESULT_DIR, "logistic_feature_coefficients.csv"),
    index=False
)

plt.figure(figsize=(10, 8))
sns.barplot(data=logistic_features.head(15), x="Coefficient", y="Feature")
plt.title("Top Logistic Regression Feature Coefficients")
plt.tight_layout()
plt.savefig(os.path.join(GRAPH_DIR, "logistic_feature_coefficients.png"))
plt.show()

# ============================================================
# 16. MODEL COMPARISON
# ============================================================

comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Logistic Regression"],
    "Task": [
        "Predict exact delivery time",
        "Classify delivery as Fast or Delayed"
    ],
    "Main Results": [
        f"MSE={mse:.4f}, RMSE={rmse:.4f}, MAE={mae:.4f}, R2={r2:.4f}",
        f"Accuracy={accuracy:.4f}, Precision={precision:.4f}, Recall={recall:.4f}, F1={f1:.4f}, ROC_AUC={roc_auc:.4f}"
    ]
})

display(comparison)

comparison.to_csv(os.path.join(RESULT_DIR, "model_comparison.csv"), index=False)

# ============================================================
# 17. FINAL REPORT
# ============================================================

insights = """
Actionable Insights:

1. Optimize delivery routes using distance and traffic data.
2. Increase delivery staff during rush hours.
3. Adjust estimated delivery time during bad weather.
4. Assign high-priority orders to experienced delivery partners.
5. Use delay prediction to notify customers early.
6. Analyse vehicle type performance in different traffic conditions.
7. Improve delivery partner training to reduce delays.
"""

final_report = f"""
# Food Delivery Time Prediction Report

## Objective

The objective of this project is to predict food delivery time using customer location, restaurant location, distance, weather, traffic, vehicle type, order priority, ratings, order cost, tip amount, and delivery partner experience.

## Dataset

Dataset Path:

{DATA_PATH}

Rows: {df.shape[0]}
Columns after processing: {df.shape[1]}

## Preprocessing Steps

- Loaded the dataset.
- Checked and handled missing values.
- Extracted latitude and longitude from customer and restaurant locations.
- Calculated distance using the Haversine formula.
- Created Order_Hour and Rush_Hour features.
- Created Delivery_Status for logistic regression.
- Encoded categorical variables using one-hot encoding.
- Handled outliers using the IQR method.
- Dropped completely empty columns before imputation.
- Applied median imputation.
- Standardized features using StandardScaler.

## Linear Regression Results

MSE: {mse:.4f}
RMSE: {rmse:.4f}
MAE: {mae:.4f}
R2 Score: {r2:.4f}

## Logistic Regression Results

Accuracy: {accuracy:.4f}
Precision: {precision:.4f}
Recall: {recall:.4f}
F1 Score: {f1:.4f}
ROC AUC: {roc_auc:.4f}

## Recommendations

{insights}

## Saved Outputs

All outputs are saved inside:

{BASE_DIR}
"""

with open(os.path.join(REPORT_DIR, "actionable_insights.txt"), "w") as f:
    f.write(insights)

with open(os.path.join(REPORT_DIR, "final_report.md"), "w", encoding="utf-8") as f:
    f.write(final_report)

with open(os.path.join(REPORT_DIR, "final_report.txt"), "w", encoding="utf-8") as f:
    f.write(final_report)

summary = {
    "Linear Regression": linear_results,
    "Logistic Regression": logistic_results,
    "ROC AUC": roc_auc,
    "Dropped Empty Columns": all_nan_cols,
    "Output Folder": BASE_DIR
}

with open(os.path.join(REPORT_DIR, "summary_results.json"), "w") as f:
    json.dump(summary, f, indent=4)

print("\nProject Completed Successfully!")
print("All results saved in:")
print(BASE_DIR)
```
